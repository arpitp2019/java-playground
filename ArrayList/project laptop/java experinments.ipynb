```java
public class Book {
    private String name;
    private String content;
    private int published;

    public Book(String name, int published, String content) {
        this.name = name;
        this.published = published;
        this.content = content;
    }

    public String getName() {
        return this.name;
    }

    public void setName(String name) {
        this.name = name;
    }

    public int getPublished() {
        return this.published;
    }

    public void setPublished(int published) {
        this.published = published;
    }

    public String getContent() {
        return this.content;
    }

    public void setContent(String content) {
        this.content = content;
    }

    public String toString() {
        return "Name: " + this.name + " (" + this.published + ")\n"
            + "Content: " + this.content;
    }
}
```

```java
HashMap<String, Book> books = new HashMap<>();
Book WoN = new Book("War of Nations", 2020, "An epic tale of war and peace.");
Book CnP = new Book("crime and punishment", 1866, "A psychological drama exploring morality and redemption.");
books.put(WoN.getName(), WoN);
books.put(CnP.getName(), CnP);
```

```java
books.get("War of Nations").toString();
```
